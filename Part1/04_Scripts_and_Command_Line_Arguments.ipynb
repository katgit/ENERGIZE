# Part 1 - Writing Python Scripts and Command-Line Arguments

So far we've worked inside a notebook. But a Python file (`.py`) can be used in two different ways:
1. **Run directly**: You execute it from the command line, e.g. `python my_script.py`.
2. **Imported as a module**: Another script uses `import my_script` to access its functions.

We need a way to tell Python: *"Only run this specific block of code when I am running the file directly."*

### The `if __name__ == "__main__":` Pattern
By convention, programmers put the main logic of their script inside a function, usually called `main()`. The `if __name__ == "__main__":` block then simply calls this function. This cleanly separates your "executable" code from your reusable "helper" functions, and is a professional way to structure Python files.

```python
import math

def calculate_circle_area(radius):
    """Calculates the area of a circle."""
    return math.pi * radius**2

def main():
    """The main logic for our script."""
    print("Calculating the area for a circle with radius 5...")
    area = calculate_circle_area(5)
    print(f"The area is: {area:.2f}")

# This ensures main() is called only when we run this file directly.
if __name__ == "__main__":
    main()
```

## Command-Line Arguments
Command-line arguments let you pass information to your script when you run it from the terminal, without editing the code. This makes your scripts flexible and automatable - you can pass in filenames, numbers, or settings on the fly.

Python's built-in `sys` module gives you access to these arguments through `sys.argv`, a list of strings:
- `sys.argv[0]` is the script name.
- `sys.argv[1]`, `sys.argv[2]`, ... are the arguments you passed in.

**Important**: All values in `sys.argv` are strings. If you need to do math, convert them first (e.g., using `int()` or `float()`).

Below is the content of `circle_calc.py`, a script that calculates the area of a circle ($A = \pi r^2$), taking the radius as a command-line argument. This code will not run inside the notebook the way it's shown (there is no command line here) - instead, save it as a `.py` file and run it from a terminal.

```python
%%writefile circle_calc.py
import sys
import math

def calculate_circle_area(radius):
    """Calculates the area of a circle given its radius."""
    return math.pi * (radius ** 2)

def main():
    """Processes command-line arguments and prints the result."""
    if len(sys.argv) < 2:
        print("Usage: python circle_calc.py <radius>")
        sys.exit()

    radius_str = sys.argv[1]
    radius = float(radius_str)

    area = calculate_circle_area(radius)
    print(f"A circle with radius {radius} has an area of {area:.2f}.")

if __name__ == "__main__":
    main()
```

The `%%writefile` cell magic above saves the cell's contents to a file called `circle_calc.py` in the current directory. Now let's run it from the terminal, right here in the notebook, using the `!` shell-escape:

```python
!python circle_calc.py 5
```

Try running it with a different radius, or with no argument at all, to see the usage message.

```python
!python circle_calc.py 12.5
```
