# Part 1 - Functions

A function is a named block of reusable code that performs a specific task.

**Why use functions?**
- **Organization**: They break your program into smaller, manageable pieces.
- **Reusability**: Write the code once and use it as many times as you need.

## Defining and Calling a Function
You define a function using the `def` keyword, followed by a name and parentheses `()`.

```python
def say_hello():
    print("Hello, world!")

# "Call" the function to run its code
say_hello()
```

## Parameters
A function can accept inputs, called **parameters** (or arguments), to make it more flexible.

```python
def greet_person(name):
    print(f"Hello, {name}! Welcome.")

greet_person("Alice")
greet_person("Bob")
```

## Return Values
A function can send a value back to the code that called it using the `return` keyword.

```python
def calculate_area(side_length):
    area = side_length * side_length
    return area

square_one_area = calculate_area(5)
square_two_area = calculate_area(10)
print(square_one_area, square_two_area)
```

## Docstrings
Good code is not just about getting the right answer - it should also be understandable to others (and to your future self!). A **docstring** is a special comment placed right after the `def` line that explains what a function does, using triple quotes `"""`.

A standard docstring includes:
- A brief, one-line summary of the function's purpose.
- A description of the input arguments (`Args:`).
- A description of the output value (`Returns:`).

```python
def convert_mpa_to_gpa(stress_mpa):
    """Converts a stress value from megapascals (MPa) to gigapascals (GPa).

    Args:
        stress_mpa (float): The stress value in megapascals.

    Returns:
        float: The equivalent stress value in gigapascals.
    """
    gpa = stress_mpa / 1000
    return gpa

yield_strength_gpa = convert_mpa_to_gpa(450.5)
print(f"The yield strength is {yield_strength_gpa} GPa.")
```

## Default Parameter Values
A function parameter can have a default value, making it optional. Optional parameters must come after any required parameters.

```python
def calculate_force(mass, acceleration=9.81):
    """Calculates force given mass and an optional acceleration.

    Defaults to calculating weight on Earth.

    Args:
        mass (float): The mass of the object in kg.
        acceleration (float): The acceleration in m/s^2 (optional).

    Returns:
        float: The calculated force in Newtons (N).
    """
    force = mass * acceleration
    return force

# Calculating weight on Earth (using the default acceleration)
weight_on_earth = calculate_force(10)
print(f"The weight of a 10kg object on Earth is {weight_on_earth:.2f} N.")

# Calculating weight on the Moon (overriding the default). Moon's gravity is about 1.62 m/s^2
weight_on_moon = calculate_force(10, 1.62)
print(f"The weight of a 10kg object on the Moon is {weight_on_moon:.2f} N.")
```
