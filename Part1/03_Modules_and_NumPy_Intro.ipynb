# Part 1 - Modules and an Introduction to NumPy

A **module** (also called a package or library) is a file containing Python code - usually functions, variables, and classes - that someone else has already written. You import modules using the `import` keyword to use their code without writing it yourself.

## The `math` Module
The `math` module provides access to fundamental mathematical functions and constants that go beyond basic `+ - * /`. It's built into Python.

```python
import math

# Use a function from the math module
root = math.sqrt(81)
print(f"The square root of 81 is: {root}")

# Use a constant from the math module
pi_value = math.pi
print(f"The value of Pi is approximately: {pi_value:.4f}")

# Calculate the area of a circle with a radius of 5
radius = 5
area = math.pi * math.pow(radius, 2)  # math.pow(x, y) is x to the power of y
print(f"The area of a circle with radius {radius} is: {area:.2f}")
```

## NumPy - The Powerhouse for Scientific Computing
For serious scientific and engineering work, we use a much more powerful module called **NumPy** (Numerical Python). The universal convention is to import it with the alias `np`.

```python
import numpy as np
```

The real power of NumPy is performing mathematical operations on an entire array at once, without writing a loop. This is called **vectorization** - it's fast and efficient, especially for large datasets common in engineering and science.

```python
# A regular Python list of measurements
stress_data_mpa = [150, 200, 250, 300, 350, 400]

# Convert the Python list into a NumPy array
stress_array = np.array(stress_data_mpa)

# Perform a calculation on the ENTIRE array at once!
stress_in_gpa = stress_array / 1000

print(f"Original Data (MPa): {stress_data_mpa}")
print(f"Converted Data (GPa): {stress_in_gpa}")
```

## Creating Arrays from Scratch
Besides converting lists, NumPy provides functions to create large arrays instantly - essential for setting up simulations or initializing data structures.

- `np.zeros(shape)`: creates an array filled with zeros.
- `np.ones(shape)`: creates an array filled with ones.
- `np.arange(start, stop, step)`: creates an array of evenly spaced values (like Python's `range` but for arrays).

```python
zero_array = np.zeros(5)
print(f"Zeros: {zero_array}")

ones_array = np.ones((2, 3))  # 2 rows, 3 columns
print(f"Ones:\n{ones_array}")

range_array = np.arange(10, 21, 2)
print(f"Range: {range_array}")
```

NumPy also provides many mathematical functions, such as trigonometric functions, that operate element-wise on arrays:

```python
angles = np.array([0, np.pi / 2, np.pi])
sines = np.sin(angles)

print(f"Angles: {angles}")
print(f"Sines: {sines}")
```

## Boolean Indexing
One of NumPy's most powerful features is selecting data based on a condition. This is called **boolean indexing** or logical filtering. You can ask "Which measurements are greater than 300?" and NumPy will give you just those values.

```python
stress_array = np.array([150, 200, 250, 300, 350, 400])

# The condition creates a boolean array (True/False)
high_stress_condition = stress_array > 300
print(f"Condition: {high_stress_condition}")

# Use this boolean array to select only the True elements
plastic_region_stresses = stress_array[high_stress_condition]
print(f"High Stress Values: {plastic_region_stresses}")
```
