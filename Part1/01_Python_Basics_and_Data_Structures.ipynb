## String Variables
Strings are sequences of characters and can be defined using single or double quotes.

```python
mat1 = "steel"
mat2 = "aluminum"
mat3 = "copper"
```

We can "add" strings together using a `+` sign:

```python
mat1 + mat2 + mat3
```

We **cannot** add numeric values to strings directly. Try running the cell below - it will raise an error:

```python
mat1 + 21
```

But we can use the multiplication symbol `*` to repeat a string several times:

```python
mat1 * 3
```

```python
# Compute length of a string
len(mat1)
```

Strings have many built-in **methods**, such as converting to upper/lower case, capitalizing, or counting characters:

```python
print(mat2.upper())
print(mat2.capitalize())
print(mat2.count('u'))
```

## Data Structures
### Lists
Python **lists** group sequences of elements. Lists can hold elements of different types and **can change** - we can add, remove, or modify values.

```python
elements = ['Al', 'Ag', 'Au', 'Co', 'Cr']
elements = elements + ['Cu', 'Fe']
print(elements)
```

Elements are accessed by position, where **0 is the index of the first element**. A negative index counts from the end.

```python
print(elements[0])
print(elements[len(elements) - 1])
print(elements[-3])
```

Use the colon `:` symbol to create a **sublist** (slice) of a list. The last index is not included.

```python
print(elements[2:4])
print(elements[:4])
print(elements[4:])
```

```python
# Find the index of a specific element
print(elements.index('Co'))

# Count how many times a value appears in a list
print(elements.count('Ag'))
```

```python
# Repeat a list, add an element, and sort it
print(elements * 2)

elements.append('Mg')
print(elements)
print(len(elements))
print(sorted(elements, reverse=True))
```

### List Comprehensions
List comprehensions offer a concise, elegant way to **create new lists based on existing lists** (or other iterable objects).

```python
squares = [x*x for x in range(6)]
print(squares)

# You can add a condition
even_squares = [x*x for x in range(6) if x % 2 == 0]
print(even_squares)
```

## Dictionaries
Dictionaries index values by a given key (**key-value pairs**).

```python
# A dictionary representing the composition of an alloy by weight percent
stainless_steel_304_composition = {
    "Fe": 69.0,  # Iron is the base element
    "Cr": 18.0,  # Chromium for corrosion resistance
    "Ni": 9.0,   # Nickel to stabilize the austenitic structure
    "Mn": 2.0,
    "Si": 1.0,
    "C": 0.08
}

print(stainless_steel_304_composition.keys())
print(stainless_steel_304_composition.values())
```

You can get a value directly using its key, without needing to know its position, as you would in a list:

```python
stainless_steel_304_composition["Ni"]
```

Keys and values do not have to be numeric, and dictionaries are **dynamic** - you can add new keys at any time:

```python
iron_properties = {
    "element_name": "Iron",
    "symbol": "Fe",
    "atomic_number": 26,
    "crystal_structure": "Body-Centered Cubic (BCC)",
    "melting_point_celsius": 1538,
    "density_g_cm3": 7.874
}

iron_properties["youngs_modulus_gpa"] = 208
print(iron_properties)
```

You can also create dictionaries within dictionaries (nested dictionaries):

```python
students = {
    "Peter": {"age": 22, "dept": "MSE"},
    "Isabel": {"age": 21, "dept": "ECE"},
    "Anna": {"age": 23, "dept": "BME"},
    "Paul": {"age": 25, "dept": "ME"},
    "Maria": {"age": 22, "dept": "SE"}
}

print(students["Anna"]["dept"])
```

## Tuples
A **tuple** is a collection of ordered items, similar to a list. The key difference is that a tuple is **immutable** - once created, it cannot be changed. Tuples are created using parentheses `()`.

```python
material_data = ("Silicon Carbide", "SiC", 4)  # (Name, Formula, Mohs Hardness)
print(material_data)
print(type(material_data))
```

You access tuple items by index, just like a list. A very common and useful feature is **unpacking**, where all the values from a tuple are assigned to separate variables in one line.

```python
# A tuple storing a measurement: (Material, Test Temperature, Result)
tensile_test_result = ("Titanium", 25, 450)

material, temperature, strength_mpa = tensile_test_result
print(f"Material: {material}")
print(f"Strength: {strength_mpa} MPa at {temperature}\u00b0C")
```

## Iteration and Loops
The most common form of iteration in Python is the `for` loop, which lets you process every item in a list (or any iterable).

```python
stress_data_mpa = [50, 150, 250, 350, 400, 450, 480, 500]

for value in stress_data_mpa:
    print(value / 1000)

print("End of loop")
```

The `range()` function is often used in `for` loops. Another common pattern is accumulating a running total, such as computing an average:

```python
for i in range(5):
    print(i * i)
```

```python
# Calculate the average age of the students in the 'students' dictionary
ave_age = 0
for student in students:
    ave_age = ave_age + students[student]["age"]

ave_age = ave_age / len(students)
print(f"The average age of students is {ave_age}")
```

### Enumerate
Sometimes you need both the index and the value while iterating over a list. Use the `enumerate()` function:

```python
mylist = [1, -5, 7, -3, 4, 2]
for i, value in enumerate(mylist):
    print(f"Index: {i}, Value: {value}")
```

## More Control Flow Tools
The `if` statement runs a block of code only if a condition is true. Think of it as: *"If it's raining, I will take an umbrella."*

```python
temperature_celsius = 32

if temperature_celsius > 30:
    print("It's a hot day!")

if temperature_celsius < 0:
    print("It's freezing!")
```

The `else` clause runs when the `if` condition is not met, and `elif` ("else if") lets you check additional conditions in order:

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  # This is the first true condition, so it runs.
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```
