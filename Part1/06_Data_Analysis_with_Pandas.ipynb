# Part 1 - Data Analysis with Pandas

While NumPy is great for raw numerical arrays, most real-world data comes with labels, columns, and mixed data types (numbers, text, dates). This is where **pandas** comes in.

Pandas is the most popular tool in Python for data manipulation and analysis. It provides a powerful object called a **DataFrame** - essentially a smart, programmable spreadsheet (or a SQL table) inside your Python script.

## Loading Data
One of the most common tasks is loading data from a file (like a CSV) into a DataFrame. Pandas makes this simple with `pd.read_csv()`. The standard convention is to import pandas with the alias `pd`.

```python
import pandas as pd

file_path = 'tensile_test_data.csv'
df = pd.read_csv(file_path)
print(df)
```

## Exploring a DataFrame
Once your data is loaded, you need to quickly understand its structure and contents:
- `.head()`: shows the first rows of the DataFrame.
- `.info()`: summarizes data types and non-null values.
- `.describe()`: calculates descriptive statistics (mean, std, min, max, etc.) for the numerical columns.

```python
print("--- .head() ---")
print(df.head(3))  # Show first 3 rows

print("\n--- .info() ---")
df.info()

print("\n--- .describe() ---")
print(df.describe())
```

## Selecting Columns
Select one column with `df['ColumnName']` (returns a pandas **Series**, a single column of a DataFrame). Select multiple columns with `df[['Column1', 'Column2']]` (note the double square brackets!).

```python
materials = df['Material']
print("--- Single Column (Series) ---")
print(materials)

results = df[['SampleID', 'PeakLoad_N']]
print("\n--- Multiple Columns (DataFrame) ---")
print(results)
```

## Creating New Columns
One of the most powerful pandas features is creating new columns from calculations on existing ones. Let's calculate the engineering stress for each sample. Assuming a cross-sectional area of 12.5 mm²:

$$\text{Stress (MPa)} = \frac{\text{PeakLoad\_N}}{\text{Area\_mm2}}$$

```python
area_mm2 = 12.5

df['Stress_MPa'] = df['PeakLoad_N'] / area_mm2
print(df)
```

## Filtering Rows
Often you only want a subset of your data meeting a certain condition. The syntax is very similar to NumPy's boolean indexing - the condition inside the brackets returns True/False for each row, and pandas returns only the rows where it was True.

```python
condition = df['Material'] == 'Steel'
steel_samples = df[condition]
print(steel_samples)
```

## Grouping Data
Pandas can group data and perform calculations on each group - the "split-apply-combine" pattern:
1. **Split**: split the data into groups (e.g., by material type).
2. **Apply**: apply a function to each group independently (e.g., calculate the mean).
3. **Combine**: combine the results into a new data structure.

```python
avg_load_by_material = df.groupby('Material')['PeakLoad_N'].mean()
print(avg_load_by_material)
```

## Sorting
Use `.sort_values()` to sort a DataFrame by one or more columns. `ascending=False` sorts from highest to lowest.

```python
strongest_first = df.sort_values(by='PeakLoad_N', ascending=False)
print(strongest_first)
```
