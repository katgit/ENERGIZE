# Part 1 - Plotting with Matplotlib

Let's create a simple line plot. The process always involves three basic steps:
1. **Import the library** - the standard convention is to import `matplotlib.pyplot` as `plt`.
2. **Create the plot** - use `plt.plot(x_data, y_data)` to generate the plot in memory.
3. **Show the plot** - use `plt.show()` to display it.

```python
import matplotlib.pyplot as plt

x_values = [0, 1, 2, 3, 4]
y_values = [0, 1, 4, 9, 16]

plt.plot(x_values, y_values)
plt.show()
```

## Labels and Titles
A plot without labels is meaningless - always tell your audience what the axes and title represent.

- `plt.title()`: sets the main title.
- `plt.xlabel()` / `plt.ylabel()`: label the axes.
- `plt.grid(True)`: adds a grid to make the plot easier to read.

```python
plt.plot(x_values, y_values)
plt.title("Relationship between X and Y")
plt.xlabel("X-Axis Values")
plt.ylabel("Y-Axis Values (X^2)")
plt.grid(True)
plt.show()
```

## Plotting with NumPy Arrays
Matplotlib works perfectly with NumPy arrays, which is how you'll handle most experimental data. Let's plot a simple stress-strain curve for a material.

```python
import numpy as np

strain = np.array([0, 0.001, 0.002, 0.003, 0.004, 0.005])
stress = np.array([0, 210, 405, 590, 720, 800])

plt.plot(strain, stress)
plt.title("Stress-Strain Curve for Steel Alloy")
plt.xlabel("Strain (mm/mm)")
plt.ylabel("Stress (MPa)")
plt.grid(True)
plt.show()
```
