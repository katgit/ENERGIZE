# Part 1 - Plotting with Matplotlib

Let's create a simple line plot. The process always involves three basic steps:
1. **Import the library** - the standard convention is to import `matplotlib.pyplot` as `plt`.
2. **Create the plot** - use `plt.plot(x_data, y_data)` to generate the plot in memory.
3. **Show the plot** - use `plt.show()` to display it.

```python
import matplotlib.pyplot as plt

x_values = [0, 1, 2, 3, 4]
y_values = [0, 1, 4, 9, 16]

plt.plot(x_values, y_values)
plt.show()
```

## Labels and Titles
A plot without labels is meaningless - always tell your audience what the axes and title represent.

- `plt.title()`: sets the main title.
- `plt.xlabel()` / `plt.ylabel()`: label the axes.
- `plt.grid(True)`: adds a grid to make the plot easier to read.

```python
plt.plot(x_values, y_values)
plt.title("Relationship between X and Y")
plt.xlabel("X-Axis Values")
plt.ylabel("Y-Axis Values (X^2)")
plt.grid(True)
plt.show()
```

## Plotting with NumPy Arrays
Matplotlib works perfectly with NumPy arrays, which is how you'll handle most experimental data. Let's plot a simple stress-strain curve for a material.

```python
import numpy as np

strain = np.array([0, 0.001, 0.002, 0.003, 0.004, 0.005])
stress = np.array([0, 210, 405, 590, 720, 800])

plt.plot(strain, stress)
plt.title("Stress-Strain Curve for Steel Alloy")
plt.xlabel("Strain (mm/mm)")
plt.ylabel("Stress (MPa)")
plt.grid(True)
plt.show()
```

## Scatter Plots
A scatter plot shows individual observations as points. It is useful for exploring the relationship between two measured variables.


```python
temperature_c = np.array([5, 10, 15, 20, 25, 30])
energy_use_kwh = np.array([42, 39, 35, 31, 28, 26])

plt.scatter(temperature_c, energy_use_kwh, color="darkorange", marker="o")
plt.title("Temperature and Energy Use")
plt.xlabel("Temperature (degrees C)")
plt.ylabel("Energy Use (kWh)")
plt.grid(True)
plt.show()
```

## Bar Charts
A bar chart compares values across discrete categories. The bar height represents the value for each category.


```python
materials = ["Steel", "Aluminum", "Copper", "Titanium"]
strength_mpa = [250, 150, 210, 300]

plt.bar(materials, strength_mpa, color=["steelblue", "darkorange", "seagreen", "firebrick"])
plt.title("Material Strength Comparison")
plt.xlabel("Material")
plt.ylabel("Strength (MPa)")
plt.show()
```

## Histograms
A histogram groups numerical data into ranges, or bins, to show its distribution. It can help reveal where values cluster and how much they vary.


```python
measurements_mpa = np.array([198, 205, 210, 212, 215, 218, 220, 221, 225, 230, 235, 240, 245, 250])

plt.hist(measurements_mpa, bins=5, color="seagreen", edgecolor="black")
plt.title("Distribution of Strength Measurements")
plt.xlabel("Strength (MPa)")
plt.ylabel("Number of Samples")
plt.show()
```

## Pie Charts
A pie chart shows how categories contribute to a whole. Use it when the parts add up to a meaningful total and there are only a few categories.


```python
energy_sources = ["Solar", "Wind", "Hydro", "Other"]
energy_share = [40, 30, 20, 10]

plt.pie(energy_share, labels=energy_sources, autopct="%1.1f%%", startangle=90)
plt.title("Renewable Energy Sources")
plt.show()
```
