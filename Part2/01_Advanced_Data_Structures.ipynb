## Sets
A **set** is an unordered collection of **unique** elements. Sets are ideal for removing duplicates and for fast membership testing (`in`).

```python
tested_materials = ['Steel', 'Aluminum', 'Steel', 'Titanium', 'Aluminum', 'Copper']

unique_materials = set(tested_materials)
print(unique_materials)
print(f"Number of distinct materials tested: {len(unique_materials)}")
```

Sets support mathematical set operations - useful for comparing groups of samples:

```python
batch_A = {'Steel', 'Aluminum', 'Titanium'}
batch_B = {'Aluminum', 'Copper', 'Titanium'}

print("In both batches (intersection):", batch_A & batch_B)
print("In either batch (union):", batch_A | batch_B)
print("Only in batch A (difference):", batch_A - batch_B)
```

## Set and Dictionary Comprehensions
Just like list comprehensions, Python supports **set comprehensions** `{...}` and **dictionary comprehensions** `{key: value ...}`.

```python
stress_data_mpa = [150, 200, 250, 300, 350, 400]

# Set comprehension: unique rounded stress categories (nearest 100 MPa)
categories = {round(s, -2) for s in stress_data_mpa}
print(categories)

# Dict comprehension: map each stress value (MPa) to its GPa equivalent
stress_lookup = {s: s / 1000 for s in stress_data_mpa}
print(stress_lookup)
```

## Sorting with a Custom Key
The built-in `sorted()` function accepts a `key` argument - a function applied to each element to determine sort order. This is essential when sorting complex objects, like a list of dictionaries.

```python
samples = [
    {"id": "A01", "material": "Steel", "peak_load_n": 45000},
    {"id": "B01", "material": "Aluminum", "peak_load_n": 28000},
    {"id": "A02", "material": "Steel", "peak_load_n": 45500},
]

# Sort samples by peak load, highest first
by_load = sorted(samples, key=lambda s: s["peak_load_n"], reverse=True)
for s in by_load:
    print(s["id"], s["peak_load_n"])
```

## `zip()` - Iterating Over Multiple Lists Together
`zip()` pairs up elements from two or more iterables, which is very handy when you have parallel lists of data (e.g., strain values and stress values from an experiment).

```python
strain = [0.000, 0.001, 0.002, 0.003]
stress = [0, 210, 405, 590]

for e, s in zip(strain, stress):
    print(f"Strain: {e:.3f}, Stress: {s} MPa")
```

## `namedtuple` - Tuples with Named Fields
Regular tuples are accessed by position (`material_data[0]`), which can be hard to read. A `namedtuple` gives you the immutability of a tuple with the readability of named attributes.

```python
from collections import namedtuple

Material = namedtuple("Material", ["name", "formula", "density_g_cm3"])

sic = Material(name="Silicon Carbide", formula="SiC", density_g_cm3=3.21)
print(sic.name, sic.formula, sic.density_g_cm3)
print(sic)
```
