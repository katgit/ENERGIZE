# Part 2 - Advanced Dictionary Usage

Dictionaries are one of the most-used data structures in real Python code - configuration files, JSON data, grouped results, and lookup tables are all dictionaries under the hood. This notebook covers patterns that go beyond basic key-value access.

## Safe Access with `.get()`
Accessing a missing key with `d[key]` raises a `KeyError`. The `.get()` method lets you provide a default value instead.

```python
density_lookup = {"Steel": 7.85, "Aluminum": 2.70, "Titanium": 4.51}

print(density_lookup.get("Steel"))
print(density_lookup.get("Gold"))              # Returns None - no error
print(density_lookup.get("Gold", "unknown"))   # Returns a custom default
```

## `collections.defaultdict` - Grouping Without Checking First
A common task is grouping items by a key, e.g. all sample IDs for each material. Normally you'd have to check "does this key already exist?" every time. `defaultdict` removes that boilerplate by automatically creating a default value (like an empty list) the first time a key is used.

```python
from collections import defaultdict

samples = [
    ("Steel", "A01"), ("Steel", "A02"), ("Aluminum", "B01"),
    ("Titanium", "C01"), ("Aluminum", "B02"),
]

samples_by_material = defaultdict(list)
for material, sample_id in samples:
    samples_by_material[material].append(sample_id)

print(dict(samples_by_material))
```

## `collections.Counter` - Counting Made Easy
`Counter` is a specialized dictionary for counting hashable items. It's perfect for tallying how many samples of each material you tested.

```python
from collections import Counter

materials_tested = ['Steel', 'Aluminum', 'Steel', 'Titanium', 'Aluminum', 'Steel']

counts = Counter(materials_tested)
print(counts)
print(counts.most_common(2))  # Two most common materials
```

## Merging Dictionaries
Python 3.9+ supports merging dictionaries with the `|` operator (values from the right-hand dictionary win on key conflicts). Older code often uses `dict(a, **b)` or `.update()`.

```python
base_properties = {"material": "Steel", "density_g_cm3": 7.85}
test_results = {"peak_load_n": 45000, "density_g_cm3": 7.90}  # Measured density overrides the reference value

combined = base_properties | test_results
print(combined)
```

## Iterating Over Dictionaries with `.items()`
When you need both the key and the value while looping, use `.items()` rather than looking up the value separately.

```python
for material, density in density_lookup.items():
    print(f"{material}: {density} g/cm^3")
```

## Nested Dictionaries and JSON
Real-world scientific data is often nested and stored as **JSON**, a text format that maps almost directly onto Python dictionaries and lists. The built-in `json` module converts between the two.

```python
import json

sample_record = {
    "sample_id": "A01",
    "material": "Steel",
    "measurements": {"peak_load_n": 45000, "elongation_mm": 10.5},
    "tags": ["batch1", "verified"],
}

# Convert a Python dict to a JSON string
json_text = json.dumps(sample_record, indent=2)
print(json_text)

# Convert a JSON string back into a Python dict
parsed = json.loads(json_text)
print(parsed["measurements"]["peak_load_n"])
```
