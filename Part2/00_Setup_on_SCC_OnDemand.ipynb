# Part 2 - Setting Up the SCC Environment

This notebook walks through the setup needed for the Part 2 workshop on Boston University's Shared Computing Cluster (SCC). You will use SCC OnDemand to access your project files, clone the workshop repository, and create a personal Python environment with the packages used in the workshop.

> **Before you begin:** Make sure you have an SCC account and have been added to the `nsf-energize` project.

## 1. Log in to SCC OnDemand

1. Open a web browser and go to [scc-ondemand.bu.edu](https://scc-ondemand.bu.edu).
2. Select the BU login option if prompted.
3. Sign in with your BU credentials and complete Duo two-factor authentication.
4. After signing in, you should see the SCC OnDemand dashboard.

SCC OnDemand provides a web-based file browser, terminal, and interactive applications such as Jupyter. It lets you work on the SCC without opening a separate SSH connection.

## 2. Create Your Workshop Folder

Use the SCC OnDemand file browser to create a personal folder in the shared workshop project directory.

1. From the dashboard, open **Files** and choose `/projectnb/nsf-energize/` directory.
2. Use **New Directory** or **New Folder** to create a folder named as your SCC login name. 
3. Open your new folder. All of your workshop files and environments should be associated with this folder.

## 3. Clone the Workshop Repository

Open a terminal from the SCC OnDemand interface. Clone the repository there:

```bash
git clone https://github.com/katgit/ENERGIZE.git
cd ENERGIZE
```

Confirm that the repository was cloned by listing its contents:

```bash
pwd
ls -l
ls -l ENERGIZE
```

You should see the `Part1`, `Part2`, and other workshop files inside the `ENERGIZE` directory.

## 4. Load Miniconda

The SCC provides software through environment modules. Load Miniconda before using `conda`:

```bash
module load miniconda
module load cuda/12.8   # To install and use the pytorch library
```

The first time you use the miniconda module, you need to set up your conda configuration. Run it only once:

```bash
setup_scc_condarc.sh
```

## 5. Create the `energize` Conda Environment

Create the environment with Python 3.14.*, Jupyter Notebook, and the scientific packages used in this workshop. The `-y` option accepts the installation prompt automatically.

```bash
conda create -n energize python=3.14 notebook jupyter ipykernel numpy pandas matplotlib scipy -y
```

Activate the environment:

```bash
conda activate energize
```

Check the Python version and confirm that the main scientific packages import correctly:

```bash
python --version
python -c "import numpy, pandas, matplotlib, scipy; print('Scientific Python packages are ready.')"
```

The Python version should be `3.14.*`. Keep the environment activated for the remaining setup commands.

## 6. Install PyTorch with CUDA Support

Install the CUDA-enabled PyTorch packages using the workshop command below:

```bash
pip3 install torch torchvision --index-url https://download.pytorch.org/whl/cu126
```

This command installs PyTorch and torchvision from the PyTorch repository for CUDA 12.6. The installation can take several minutes.

Verify the installation:

```bash
python -c "import torch; print('PyTorch version:', torch.__version__); print('CUDA available:', torch.cuda.is_available())"
```

`CUDA available: True` requires that the command is run on a compute node with a compatible GPU. It may be `False` on an SCC login node even when PyTorch is installed correctly.

## 7. Run an Interactive Desktop Session

Use an interactive Desktop session when you want to work in the graphical VS Code interface on the SCC.

1. Return to the SCC OnDemand dashboard and select **Interactive Apps > Desktop**.
2. Set the **Working Directory** to your personal ENERGIZE folder, for example:
   `/projectnb/nsf-energize/YOUR_LOGIN_NAME/ENERGIZE`
3. Set the session duration to **12 hours**.
4. If your work requires a GPU, request **1 GPU**. Otherwise, leave the GPU request at zero.
5. Submit the request and wait for the Desktop session to open.

Once the Desktop session opens, open a terminal window and run:

```bash
module load miniconda
module load cuda/12.8
module load vscode
conda activate energize
export LD_LIBRARY_PATH="$CONDA_PREFIX/lib:${LD_LIBRARY_PATH:-}"
code
```

When VS Code opens:

1. Open the `ENERGIZE` folder, if it is not already open.
2. Open a notebook from the `Part2` folder.
3. If needed, select the **energize** kernel using the kernel selector in the bottom-right corner of VS Code.
4. If VS Code prompts you to install extensions, install the necessary **Python** and **Jupyter** extensions published by Microsoft.

The selected `energize` kernel should be used when running the workshop notebooks.
